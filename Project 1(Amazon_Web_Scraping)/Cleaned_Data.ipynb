```python
import pandas as pd
```

```python
df = pd.read_csv(r"C:\Users\Abdul Sofiyan\OneDrive\Desktop\Project1(Web Scraping)\Amazon_laptops.csv", on_bad_lines="skip")
df
```

```python
print(df.shape)
print(df.info())

# Check missing values
print(df.isnull().sum())

# Check duplicate rows
print("Duplicate rows:", df.duplicated().sum())

#  Remove duplicate rows
df = df.drop_duplicates()


#  Remove extra spaces from column names
df.columns = df.columns.str.strip()

#  Remove extra spaces from text columns
for col in df.select_dtypes(include="object"):
    df[col] = df[col].str.strip()
```

```python
# Remove duplicate rows
df = df.drop_duplicates()

#  Remove extra spaces from column names
df.columns = df.columns.str.strip()

#  Remove extra spaces from text columns
for col in df.select_dtypes(include="object"):
    df[col] = df[col].str.strip()

#  Check missing values again
print(df.isnull().sum())

#  Fill missing numerical values with median
for col in df.select_dtypes(include="number"):
    df[col] = df[col].fillna(df[col].median())

#  Fill missing categorical values
for col in df.select_dtypes(include="object"):
    df[col] = df[col].fillna("Unknown")

#  Final check
print(df.info())
print(df.isnull().sum())
print(df.duplicated().sum())
```

```python
# using title() for the columns headings
df.columns = df.columns.str.strip().str.title()
df.columns
```

```python
# Converting from text to numeric, removing ₹.
# Converting text prices to numeric
df["Price"] = (
    df["Price"]
    .astype(str)
    .str.replace("₹", "", regex=False)
    .str.replace(",", "", regex=False)
    .str.strip()
)

df["Price"] = pd.to_numeric(df["Price"], errors="coerce")
df["Price"]
```

```python
# Handling missing ratings and checking invalid values
df["Rating"] = df["Rating"].isnull().sum()
df["Rating"] = pd.to_numeric(df["Rating"], errors="coerce")
df["Rating"]
```

```python
# Removing spaces and standardize capitalization in Brand column
df["Brand"] = df["Brand"].str.strip().str.title()
df["Brand"]
```

```python
df["Ram"] = pd.to_numeric(df["Ram"], errors="coerce")
df["Ram"] = df["Ram"].fillna(df["Ram"].median())

df["Ram"]
```

```python
# Extract RAM number, e.g. 16GB → 16 and 1TB to 1064
ram = df["Ram"].astype(str)

df["Ram"] = ram.str.extract(r"(\d+\.?\d*)")[0].astype(float)

df.loc[ram.str.contains("TB", case=False, na=False), "Ram"] *= 1064

df["Ram"] = df["Ram"].bfill().ffill()

df["Ram"]
```
