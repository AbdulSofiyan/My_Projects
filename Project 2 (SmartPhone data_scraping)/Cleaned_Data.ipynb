```python
import pandas as pd
```

```python
df = pd.read_csv(r"C:\Users\Abdul Sofiyan\Downloads\archive\Mobiles Dataset (2025).csv",encoding = "latin1")
df
```

```python
print(df.shape)
print(df.info())

# Check missing values
print(df.isnull().sum())

# Check duplicate rows
print("Duplicate rows:", df.duplicated().sum())

#  Remove duplicate rows
df = df.drop_duplicates()


#  Remove extra spaces from column names
df.columns = df.columns.str.strip()

#  Remove extra spaces from text columns
for col in df.select_dtypes(include="object"):
    df[col] = df[col].str.strip()
```

```python
# Remove duplicate rows
df = df.drop_duplicates()

#  Remove extra spaces from column names
df.columns = df.columns.str.strip()

#  Remove extra spaces from text columns
for col in df.select_dtypes(include="object"):
    df[col] = df[col].str.strip()

#  Check missing values again
print(df.isnull().sum())

#  Fill missing numerical values with median
for col in df.select_dtypes(include="number"):
    df[col] = df[col].fillna(df[col].median())

#  Fill missing categorical values
for col in df.select_dtypes(include="object"):
    df[col] = df[col].fillna("Unknown")

#  Final check
print(df.info())
print(df.isnull().sum())
print(df.duplicated().sum())
```

```python
# converting mobile weight to to number
df["Mobile Weight"] = ( 
  df["Mobile Weight"].str.replace("g", "", regex=False).str.strip()
)
df["Mobile Weight"] = pd.to_numeric(
    df["Mobile Weight"], errors="coerce"
)
```

```python
df["Mobile Weight"]
```

```python
# Cleaning Ram column
df["RAM"] = df["RAM"].replace(
    ["LPDDR4", "LPDDR5", "DDR5", "DDR7", "RAM"],
    None
)

df["RAM"] = df["RAM"].str.replace("GB", "", regex=False)

df["RAM"] = pd.to_numeric(df["RAM"], errors="coerce")

df["RAM"]
```

```python
# cleaning the Front camera column
df["Front Camera MP"] = pd.to_numeric(
df["Front Camera"].str.replace("MP", "", regex=False),errors="coerce"
)
df["Front Camera MP"]
```

```python
# cleaning the back camera column
df["Back Camera MP"] = pd.to_numeric(
df["Back Camera"].str.replace("MP", "", regex=False),errors="coerce"
)
df["Back Camera MP"]
```

```python
# cleaning batter column
df["Battery Capacity"] = (
    df["Battery Capacity"].astype(str)
    .str.replace(",", "", regex=False)
    .str.replace("mAh", "", regex=False)
    .str.strip()
)
df["Battery Capacity"] = pd.to_numeric(
    df["Battery Capacity"],
    errors="coerce"
)
df["Battery Capacity"]
```

```python
# clean screen size
df["Screen Size"] = (
    df["Screen Size"]
    .str.replace("inches", "", regex=False)
    .str.strip()
)

df["Screen Size"] = pd.to_numeric(
    df["Screen Size"],
    errors="coerce"
)
df["Screen Size"]
```

```python
# cleaning price column
price_columns = [
    "Launched Price (Pakistan)",
    "Launched Price (India)",
    "Launched Price (China)",
    "Launched Price (USA)",
    "Launched Price (Dubai)"
]

for col in price_columns:
    df[col] = (
        df[col].astype("string").str.replace(r"[^\d.]", "", regex=True)
    )

    df[col] = pd.to_numeric(
        df[col],
        errors="coerce"
    )

    
```
