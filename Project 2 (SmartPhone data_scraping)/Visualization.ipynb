```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
```

```python
df = pd.read_csv(r"C:\Project 2 (SmartPhone data_scraping)\Mobiles_Dataset_Cleaned.csv")
```

# 1. Top 10 Mobile Companies by Number of Models

```python
plt.figure(figsize=(10,5))
df["Company Name"].value_counts().head(10).plot(kind="bar")
plt.title("Top 10 Mobile Companies by Number of Models")
plt.xlabel("Company")
plt.ylabel("Number of Models")
plt.xticks(rotation=45)
plt.show()
```

# 2. RAM Distribution

```python
plt.figure(figsize=(8,5))
sns.histplot(df["RAM"], bins=10, kde=True)
plt.title("RAM Distribution")
plt.xlabel("RAM (GB)")
plt.ylabel("Number of Mobiles")
plt.show()
```

# 3. Battery Capacity Distribution

```python
plt.figure(figsize=(8,5))
sns.histplot(df["Battery Capacity"], bins=15, kde=True)
plt.title("Battery Capacity Distribution")
plt.xlabel("Battery Capacity (mAh)")
plt.ylabel("Number of Mobiles")
plt.show()
```

# 4. Screen Size Distribution

```python
plt.figure(figsize=(8,5))
sns.histplot(df["Screen Size"], bins=15, kde=True)
plt.title("Screen Size Distribution")
plt.xlabel("Screen Size (inches)")
plt.ylabel("Number of Mobiles")
plt.show()
```

# 5. Average RAM by Company

```python
top_companies = df["Company Name"].value_counts().head(10).index

temp = df[df["Company Name"].isin(top_companies)]

plt.figure(figsize=(10,5))
sns.barplot(data=temp, x="Company Name", y="RAM", estimator="mean")
plt.title("Average RAM by Company")
plt.xlabel("Company")
plt.ylabel("Average RAM (GB)")
plt.xticks(rotation=45)
plt.show()
```

# 6. Battery Capacity vs RAM

```python
plt.figure(figsize=(8,5))
sns.scatterplot(
    data=df,
    x="RAM",
    y="Battery Capacity"
)
plt.title("RAM vs Battery Capacity")
plt.xlabel("RAM (GB)")
plt.ylabel("Battery Capacity (mAh)")
plt.show()
```

# 7. Weight Distribution

```python
plt.figure(figsize=(8,5))
sns.histplot(df["Mobile Weight"], bins=20, kde=True)
plt.title("Mobile Weight Distribution")
plt.xlabel("Weight (g)")
plt.ylabel("Number of Mobiles")
plt.show()
```

# 8. Mobiles Launched by Year

```python
year_count = df["Launched Year"].value_counts().sort_index()

plt.figure(figsize=(10,5))
plt.plot(year_count.index, year_count.values, marker="o")
plt.title("Number of Mobiles Launched by Year")
plt.xlabel("Year")
plt.ylabel("Number of Mobiles")
plt.grid()
plt.show()
```
