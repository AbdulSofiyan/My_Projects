```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv(r"C:\Project 3(Advanced Multi-Source Data Pipeline)\4.1_cleaned_product_data.csv")
df.head()
```

# 1. Univariate Analysis

```python
# Book Price Distribution

plt.figure(figsize=(8, 5))

sns.histplot(df["Price"], kde=True)

plt.title("Distribution of Book Prices")
plt.xlabel("Price")
plt.ylabel("Frequency")

plt.show()
```

# Bivariate Analysis

```python
# Rating vs Price relationship (Books)

plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=df,
    x="Rating",
    y="Price"
)

plt.title("Rating vs Book Price")
plt.xlabel("Rating")
plt.ylabel("Price")

plt.show()
```

```python
# Correlation heatmap (numerical features)

numeric_df = df[["Price", "Rating", "Stock"]]

# Calculate correlation
correlation = numeric_df.corr()

# Heatmap
plt.figure(figsize=(8, 6))

sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Correlation Heatmap - Books Dataset")
plt.show()
```

```python
# Rating vs Price relationship (Books)
plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=df,
    x="Rating",
    y="Price"
)

plt.title("Rating vs Book Price")
plt.xlabel("Rating")
plt.ylabel("Price")
plt.show()
```

```python
# Scatter Plot with Trend Line
plt.figure(figsize=(8, 5))

sns.regplot(
    data=df,
    x="Rating",
    y="Price"
)

plt.title("Rating vs Book Price with Trend Line")
plt.xlabel("Rating")
plt.ylabel("Price")
plt.show()
```

# Multivariate Analysis

```python
plt.figure(figsize=(8, 5))

sns.boxplot(
    data=df,
    x="Price_Range",
    y="Rating"
)

plt.title("Price Range vs Book Rating")
plt.xlabel("Price Range")
plt.ylabel("Rating")
plt.xticks(rotation=45)

plt.show()
```
