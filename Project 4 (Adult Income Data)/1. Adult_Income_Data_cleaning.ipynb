```python
import pandas as pd
import numpy as np
```

```python
columns = [
    "age",
    "workclass",
    "fnlwgt",
    "education",
    "education_num",
    "marital_status",
    "occupation",
    "relationship",
    "race",
    "sex",
    "capital_gain",
    "capital_loss",
    "hours_per_week",
    "native_country",
    "income"
]
```

```python
df_data = pd.read_csv(
    r"C:\All project\adult python project\adult.data",names=columns,skipinitialspace=True
)
```

```python
df_test = pd.read_csv(
    r"C:\All project\adult python project\adult.test",names=columns,skipinitialspace=True
    ,skiprows=1
)
```

```python
df_data["source_split"]= "data"
df_test["source_split"]= "test"
```

```python
df = pd.concat(
    [df_data, df_test],
    ignore_index=True
)
df
```

```python
# replace ? with nan

df.replace("?", np.nan)
```

```python
categorical_columns = df.select_dtypes(include="object").columns
for col in categorical_columns:
    df[col]=df[col].str.strip()

df["workclass"].head(10).tolist()
```

```python
# check in there are no leading or trailing space
for col in df.select_dtypes(include="object").columns:
    print(col,df[col].str.contains(r"^\s|\s$",regex=True, na=False).sum())
    
```

```python
df = df.replace("?",np.nan)

missind_cols = [
    "workclass",
    "occupation",
    "native_country"
]

for col in missind_cols:
    df[col] = df[col].fillna("Unknown")

df
```

```python
numeric_col =[
    "age",
    "fnlwgt",
    "education_num",
    "capital_gain",
    "capital_loss",
    "hours_per_week"
]

for col in numeric_col:
    df[col] = pd.to_numeric(df[col],errors="coerce")

df[numeric_col].dtypes 
```

```python
df.duplicated().sum()
```

```python
df.head()
```

```python
df["income"]=df["income"].str.replace(".","",regex=False)
df
```

```python
# create a numerical target (1>50k or 0<=50k)
df["high_income"] = (df["income"] == ">50K").astype(int)
df[["income","high_income"]].value_counts()
```

```python
# create a column age group

bins = [0,25,35,45,55,65,100]

labels = [
    "18-25",
    "26-35",
    "36-45",
    "46-55",
    "56-65",
    "66-100"
]
df["age_group"]=pd.cut(df["age"],bins=bins,labels=labels)
df
```

```python
#  create a column working hour group
bin = [0,20,40,60,100]
labels = [
    "0-20",
    "21-40",
    "41-60",
    "61+"
]
df["hours_group"]=pd.cut(df["hours_per_week"],bins=bin,labels=labels)
df
```

```python
# create a column net capital gain/loss

df["net_capital"]=(df["capital_gain"]-df["capital_loss"])
df
```
