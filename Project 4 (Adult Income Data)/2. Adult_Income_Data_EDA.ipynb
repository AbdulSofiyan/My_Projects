```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv(r"C:\All project\adult python project\adult_income_data_cleaned.csv")
df
```

```python
# income distribution
```

```python
# income distribution graph

plt.figure(figsize=(8, 6))
sns.countplot(x='income', data=df)
plt.title('Income Distribution')
plt.xlabel('Income')
plt.ylabel('Number of People')
plt.show()
print("-"*40)
print(f"income count: {df['income'].value_counts()}")
```

```python
# Percentage of people earning above $>50K
percentage_high_income = (df['income'] == '>50K').sum() / len(df) * 100
print(f"Percentage of people earning above $50K: {percentage_high_income:.2f}%")
```

```python
# Education VS Income
```

```python
education_income = (df.groupby('education')['high_income'].mean().sort_values(ascending=False))
print(education_income)
print("-"*40)
education_income_pertancage = education_income*100
print(education_income_pertancage)       
```

```python
plt.figure(figsize=(10,6))

education_income_pertancage.plot(kind='bar')
plt.title("High Income Rate By Eduation")
plt.xlabel("Eduation")
plt.ylabel("High Income Rate %")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# Occupation vs Income
```

```python
plt.figure(figsize=(10,6))
sns.countplot(data=df,x="occupation",hue="income")
plt.title("Income by Occupation")
plt.xlabel("Occupation")
plt.ylabel("Number of people")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# Age Group vs Income
```

```python
plt.figure(figsize=(10,6))
sns.countplot(data=df,x="age_group",hue="income")
plt.title("Income by age group")
plt.xlabel("age group")
plt.ylabel("Number of people")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
# Workclass vs Income
```

```python
plt.figure(figsize=(10,6))
sns.countplot(data=df,x="workclass",hue="income")
plt.title("Income by Workclass")
plt.xlabel("Workclass")
plt.ylabel("Number of people")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
# Sex vs Income
```

```python
plt.figure(figsize=(10,6))
sns.countplot(data=df,x="sex",hue="income")
plt.title("Income by sex")
plt.xlabel("gender")
plt.ylabel("Number of people")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
# Weekly Working Hours VS Income
```

```python
plt.figure(figsize=(10,6))
sns.boxplot(data=df,y="hours_per_week",x="income")
plt.title("Weekly Working Hours by Income")
plt.xlabel("Income")
plt.ylabel("Hours per Week")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()

print("-"*50)
print(df.groupby("income")["hours_per_week"].mean())
print("-"*50)
print(df.groupby("income")["hours_per_week"].median())
```

```python
# Analyze Capital Gain
```

```python
print(df["capital_gain"].describe)
print("-"*60)
sns.boxplot(data=df,x="income",y="capital_gain")
plt.title("Capital Gain by Income")
plt.show()
```

```python
# Check Relationship B/W Numeric Columns
```

```python
numeric_cols =[
    "age",
    "fnlwgt",
    "education_num",
    "capital_gain",
    "capital_loss",
    "hours_per_week"
]
df[numeric_cols].corr()
plt.figure(figsize=(10,6))

sns.heatmap(df[numeric_cols].corr(),annot=True)
plt.title("Correlation Between Numeric Variables")
plt.show()
```
