```python
print(df[['Start_Station_Id', 'End_Station_Id']].dtypes)

print(df5[['Station Id']].dtypes)

print(df['Start_Time'].dtype)
print(df4['Time'].dtype)
```

```python
df5['Station Id'] = pd.to_numeric(
    df5['Station Id'],
    errors='coerce'
).astype('Int64')
```

```python
df['Start_Station_Id'] = df['Start_Station_Id'].astype('Int64')
df['End_Station_Id'] = df['End_Station_Id'].astype('Int64')
```

```python
# Merging the station data

# Start station
start_station = df5.rename(columns={
    'Station Id': 'Start_Station_Id',
    'Name': 'Station_Name_Start',
    'Longitude': 'Station_Longitude_Start',
    'Latitude': 'Station_Latitude_Start',
    'Capacity': 'Station_Capacity_Start',
    'Has Kiosk': 'Has_Kiosk_Start',
    'Station Type': 'Station_Type_Start'
})

merged_df = df.merge(
    start_station,
    on='Start_Station_Id',
    how='left'
)
```

```python
# End Station
end_station = df5.rename(columns={
    'Station Id': 'End_Station_Id',
    'Name': 'Station_Name_End',
    'Longitude': 'Station_Longitude_End',
    'Latitude': 'Station_Latitude_End',
    'Capacity': 'Station_Capacity_End',
    'Has Kiosk': 'Has_Kiosk_End',
    'Station Type': 'Station_Type_End'
})

merged_df = merged_df.merge(
    end_station,
    on='End_Station_Id',
    how='left'
)
```

```python
# Checking the merged data
print(merged_df.shape)
```

```python
# Merging the weather data

# Creating the hourly key
merged_df['Weather_Time'] = merged_df['Start_Time'].dt.floor('h')
```

```python
final_df = merged_df.merge(
    df4,
    left_on='Weather_Time',
    right_on='Time',
    how='left'
)
```

```python
print(final_df.shape)
print(final_df.head())
```

```python
station_columns_to_remove = [
    'Station_Name_Start',
    'Station_Longitude_Start',
    'Station_Latitude_Start',
    'Station_Capacity_Start',
    'Has_Kiosk_Start',
    'Station_Type_Start',
    'Station_Name_End',
    'Station_Longitude_End',
    'Station_Latitude_End',
    'Station_Capacity_End',
    'Has_Kiosk_End',
    'Station_Type_End'
]

final_df = final_df.drop(columns=station_columns_to_remove)
```

```python
print(final_df.shape)
print(final_df.columns)
```
