```python
import pandas as pd
```

```python
fe = pd.read_csv(r"C:\Project 5 (Citi Bike Project)\1.1_CitiBike_Final_Cleaned_Data.csv")
fe
```

```python
fe["Start_Time"] = pd.to_datetime(fe["Start_Time"])
fe["Stop_Time"] = pd.to_datetime(fe["Stop_Time"])
fe["Weather_Time"] = pd.to_datetime(fe["Weather_Time"])
fe["Time"] = pd.to_datetime(fe["Time"])
```

```python
# Time-based features

fe["Start_Hour"] = fe["Start_Time"].dt.hour
fe
```

```python
fe["Day_of_Week"] = fe["Start_Time"].dt.day_name()
fe
```

```python
# What day of the week does each trip start

fe["Day_Type"] = fe["Start_Time"].dt.dayofweek.map(
    lambda x: "Weekend" if x >= 5 else "Weekday"
)
fe
```

```python
# Is the trip taken on a weekday or weekend 

fe["Month"] = fe["Start_Time"].dt.month
fe
```

```python
# Peak-hour feature

fe["Peak_Hour"] = fe["Start_Hour"].isin(
    [7, 8, 9, 17, 18, 19]
)
fe
```

```python
fe["Season"] = fe["Month"].map({
    1: "Winter",
    2: "Winter",
    3: "Spring"
})
fe
```

```python
fe.to_csv("2.1_Advanced_Clean_data.csv",index = True)
```
